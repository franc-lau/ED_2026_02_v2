```python
def MinMax(L):
    if len(L)==1:
        min=L[0]
        max=L[0]
        return min, max
    min, max= MinMax(L[:-1])
    a=L[-1]
    if a<min:
        min= a
    if a>max:
        max= a
    return min,max
```
