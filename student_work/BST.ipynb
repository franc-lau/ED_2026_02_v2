```python
def find_min(T:LinkedBinaryTree):
    """Retorna el contenedor donde se encuentra el valor mínimo."""
    current= T.root()
    while T.left(current) is not None:
        current = T.left(current)
    return current
```

```python
def find_max(T:LinkedBinaryTree):
    """Retorna el contenedor donde se encuentra el valor maximo."""
    current= T.root()
    while T.right(current) is not None:
        current = T.right(current)
    return current
```
